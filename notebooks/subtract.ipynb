```python
def subtract(a, b):
    return a - b
```

```python
def distance_from_zero(a,b):
    return abs(subtract(a,b))
```

```python
print(subtract(10, 5)) 
print(subtract(20, 8)) 
print(subtract(12.5, 3.2)) 
```

```python
print(distance_from_zero(3,12))
print(distance_from_zero(0,12))
print(distance_from_zero(3,-3))
```
