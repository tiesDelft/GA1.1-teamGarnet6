```python
import math
def multiply(a, b):
    return a * b
```

```python
print(multiply(3, 4))
print(multiply(5, 6))
print(multiply(2.5, 4.0))
```

```python
# extension function to square a number 
def square(c):
    return multiply(c, c)
```

```python
print(square(3))
print(square(5.2))
print(square(0))
```
